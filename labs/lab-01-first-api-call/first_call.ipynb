```python
from dotenv import load_dotenv
import os
load_dotenv()
from openai import OpenAI
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
try:
  response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
      {"role": "system", "content": "You are a helpful, concise assistant."},
      {"role": "user", "content": "Answer in one sentence: What is the capital of Argentina?"}
    ],
    response_format={
      "type": "text"
    },
    temperature=1,
    # temperature=0.3  # More deterministic for factual answers
    max_completion_tokens=2048,
    top_p=1,
    frequency_penalty=0,
    presence_penalty=0,
  )
  print(response.choices[0].message.content)
except Exception as e:
    print(f"Error: {e}")
```

*model*: Specifies which AI model to use (gpt-4.1-mini)

*messages*: An array of conversation messages. Each has a role (system/user/assistant) and content

*response_format*: Set to "text" for plain text responses (also supports JSON)

*temperature* =1: Controls randomness (0-2 scale). 1 is balanced; lower = more focused, higher = more creative

*max_completion_tokens*=2048: Maximum length of the response

*top_p=1*: Alternative to temperature for controlling randomness (nucleus sampling)

*frequency_penalty*=0: Reduces repetition of tokens (-2 to 2). 0 = no penalty

*presence_penalty*=0: Encourages new topics (-2 to 2). 0 = no penalty
